# Fashion Deep Learning Recommendation Systems

This project is a Streamlit-powered fashion recommender system that leverages a pre-trained ResNet50 model to extract deep visual features and uses k-Nearest Neighbors with cosine similarity to suggest visually matching products.

<img src='https://aaft.com/blog/wp-content/uploads/2024/02/Fashion-Future-with-AI.jpg' width='850'>

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("paramaggarwal/fashion-product-images-small")

print("Path to dataset files:", path)
```

```python
import numpy as np
import pickle as pkl
import tensorflow as tf
from tensorflow.keras.applications.resnet50 import ResNet50,preprocess_input
from tensorflow.keras.preprocessing import image
from tensorflow.keras.layers import GlobalMaxPool2D

from sklearn.neighbors import NearestNeighbors
import os
from numpy.linalg import norm
```

```python
# Define the images directory using the 'path' variable from kagglehub
images_dir = os.path.join(path, 'images')

filenames = []
for file in os.listdir(images_dir):
    filenames.append(os.path.join(images_dir, file))
```

```python
len(filenames)
```

```python
#Importing ResNet50 Model and Cofiguration
model = ResNet50(weights='imagenet', include_top=False, input_shape=(224,224,3))
model.trainable = False

model = tf.keras.models.Sequential([model,
                                   GlobalMaxPool2D()
                                   ])
model.summary()
```

```python
#Extracting Fetaures from Image
def extract_features_from_images(image_path, model):
    img = image.load_img(image_path, target_size=(224,224))
    img_array = image.img_to_array(img)
    img_expand_dim = np.expand_dims(img_array, axis=0)
    img_preprocess = preprocess_input(img_expand_dim)
    result = model.predict(img_preprocess).flatten()
    norm_result = result/norm(result)
    return norm_result
```

```python
extract_features_from_images(filenames[0], model)
```

```python
image_features = []
for file in filenames[0:5]:
    image_features.append(extract_features_from_images(file, model))
image_features
```

```python
Image_features = pkl.dump(image_features, open('Images_features.pkl','wb'))
```

```python
filenames = pkl.dump(filenames, open('filenames.pkl','wb'))
```

## Conclusion

In this project, I built a content-based fashion recommender system that suggests visually similar products using deep learning features and nearest-neighbor search.

**What was done**
- Loaded the Fashion Product Images (Small) dataset from Kaggle (44,441 product images).
- Used a pre-trained **ResNet50** (ImageNet weights, frozen) with global max pooling to turn each image into a 2048-dimensional feature vector.
- L2-normalized every vector, so Euclidean distance ranks products the same way cosine similarity does.
- Fitted a **k-Nearest Neighbors** model (k=5, brute force) and retrieved the most similar items for a query image.

**Key takeaways**
- Transfer learning removes the need to train a model from scratch: pre-trained visual features are enough to find similar-looking products without any labels.
- The full pipeline works end to end: feature extraction, saving features with pickle, similarity search and displaying the recommendations.

**Limitations**
- For this demo, features were extracted for only 5 images to validate the pipeline. Recommendations over the full catalog require extracting all 44,441 images.
- There is no quantitative evaluation yet (for example precision@k against product categories); results were checked visually.
- Similarity is purely visual. It ignores category, color preferences, price and user behavior.
